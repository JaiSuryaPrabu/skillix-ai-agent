```python
#import ADK components
from typing import Any, Dict

from google.adk.agents import LlmAgent
from google.adk.models.google_llm import Gemini
from google.genai import types

print("✅ ADK components imported successfully.")
```

```python
#Configure Retry Options
retry_config = types.HttpRetryOptions(
    attempts=5,  # Maximum retry attempts
    exp_base=7,  # Delay multiplier
    initial_delay=1,
    http_status_codes=[429, 500, 503, 504],  # Retry on these HTTP errors
)
```

```python
planning_agent = LlmAgent(
    model=Gemini(model="gemini-2.5-flash-lite", retry_options=retry_config),
    name="PlanningAgent",
    description="Builds personalized learning syllabus based on user context.",
    instruction="""You are the Planning Agent, responsible for generating a customized learning roadmap.

You will receive:
- Topic the user wants to learn
- Current skill level (Beginner/Intermediate/Advanced)
- Learning style (theory-first, application-first, hybrid)
- Knowledge base gathered by the Context Agent
Your task:
- Analyze the learner profile + Knowledge Base.
- Generate a multi-week, structured syllabus.
- Ensure the roadmap reflects the user’s style:
- Theory-first: emphasize conceptual and foundational understanding
- Application-first: emphasize projects, exercises, hands-on practice
- Hybrid: balance both theory and application.
- Reference the knowledge base when appropriate to choose learning materials.
- Deliver a clear, step-by-step weekly plan plus a short summary explaining your reasoning.""",
    output_key="output_plan", 
)

print("✅ planning_agent created.")
```
