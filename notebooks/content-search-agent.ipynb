```python
#import ADK components
from typing import Any, Dict

from google.adk.agents import LlmAgent
from google.adk.models.google_llm import Gemini
from google.adk.tools import google_search
from google.genai import types

print("✅ ADK components imported successfully.")
```

```python
#Configure Retry Options
retry_config = types.HttpRetryOptions(
    attempts=5,  # Maximum retry attempts
    exp_base=7,  # Delay multiplier
    initial_delay=1,
    http_status_codes=[429, 500, 503, 504],  # Retry on these HTTP errors
)
```

```python
# Content Search Agent: Its job is to use the google_search tool and present content.
content_search_agent = LlmAgent(
    model=Gemini(model="gemini-2.5-flash-lite", retry_options=retry_config),
    name="ContentSearchAgent",
    description="Agent to search the internet for information.",
    instruction="""You are an expert content search agent.
Task:
- Read the provided topic.
- Your only job is to use the google_search tool to find 2-3 pieces of relevant information on the given topic
- Write a concise explanation in simple language suitable for students.
- Include 2-3 clear examples illustrating the concept.
- Keep the explanation under 180 words.
- Use short sentences and simple language.
- Do not include unnecessary details, long derivations, or unrelated content.""",
    tools=[
        google_search,
    ],
    output_key="content_findings", # The result of this agent will be stored in the session state with this key.
)

print("✅ content_agent created.")
```
